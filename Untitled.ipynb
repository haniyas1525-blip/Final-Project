```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("ecommerce_sales_34500.csv")

df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df.head()
```

```python
conn = sqlite3.connect("sales_analysis.db")
```

```python
df.to_sql("online_sales", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT *
FROM online_sales
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    strftime('%Y', order_date) AS year,
    strftime('%m', order_date) AS month,
    SUM(total_amount) AS monthly_revenue,
    COUNT(*) AS order_volume
FROM online_sales
GROUP BY year, month
ORDER BY year, month;
"""

monthly_sales = pd.read_sql_query(query, conn)

monthly_sales
```

```python
query = """
SELECT
    strftime('%Y', order_date) AS year,
    strftime('%m', order_date) AS month,
    SUM(total_amount) AS monthly_revenue,
    COUNT(*) AS order_volume
FROM online_sales
GROUP BY year, month
ORDER BY year, month;
"""

monthly_sales = pd.read_sql_query(query, conn)

monthly_sales
```

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS month,
    SUM(total_amount) AS revenue
FROM online_sales
GROUP BY month
ORDER BY month;
"""

monthly_revenue = pd.read_sql_query(query, conn)

monthly_revenue
```

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS month,
    COUNT(*) AS order_volume
FROM online_sales
GROUP BY month
ORDER BY month;
"""

monthly_orders = pd.read_sql_query(query, conn)

monthly_orders
```

```python
query = """
SELECT
    SUM(total_amount) AS total_revenue
FROM online_sales;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS total_orders
FROM online_sales;
"""

pd.read_sql_query(query, conn)
```
